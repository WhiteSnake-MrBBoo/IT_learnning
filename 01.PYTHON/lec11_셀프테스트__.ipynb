# **제어문 : for**

* **[암기] 1부터 5까지 for문 사용해 출력**

````python
for v in [1,2,3,4,5] :
    print(v, end="\t")
````

````python
for v in range(1,6) :
    print(v, end="\t")
````

* **[활용] 1부터 5까지 for문 사용해 총합 출력**

````python
total = 0
for v in range(1,6) :
    total = total + v
   
print(total)
````

* **[활용] 1부터 10까지 for문 사용해 짝수 총합 출력**

````python
total = 0
for v in range(1,11) :
    if (v%2 == 0) :
        total = total + v
   
print(total)
````

* **[활용] 1부터 10까지 for문 사용해 짝수만 출력**

````python
for v in range(1,11) :
    if (v%2 == 0) :
        print(v, end="\t")
````

# **제어문 : while**

* **[암기] 1부터 5까지 while문 사용해 출력**

````python
num = 1
while(num<=5) :
    print(num, end="\t")
    num = num + 1
````

````python
num = 1
while(True) :
    if num > 5 :
        break
    else :
        print(num, end="\t")
        num = num + 1
````

* **숫자를 입력받아 10배 값으로 반환하는 함수**
    * 함수이름 : calc_point
    * 파라미터 : num
    * 기능 : 입력값의 10배
    * 리턴 : 10배값

````python
def calc_point(num)     :    
    res = num * 10
    return res      
    
a= calc_point(6)  
print(a)
````

* **숫자 2개를 입력받아 합의 값을 돌려주는 함수**

    * 파라미터 : num1, num2
    * 기능 : num1 2배값과 num2 10배값의 합
    * 리턴 : 계산값

````python
def calc_point2(num1, num2)     :  
    res = num1*2 + num2*10
    return res

a= calc_point2(4, 5) 
print(a)
````
