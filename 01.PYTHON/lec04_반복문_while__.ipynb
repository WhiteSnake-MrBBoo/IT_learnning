````python
number = 1
while (True) :
    print(number)
    if (number == 5) : 
        break
    number = number + 1
````

````python
number = 1
while (number<=5) :
    print(number)
    number = number + 1
````

````python
dan = 2
while (dan <=9) :
    print(f"{dan}단", end="\t")
    #dan = dan + 1
    dan += 1
````
