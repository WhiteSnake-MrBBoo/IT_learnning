### **C (Create)**

```python
mylist = [1,2,3,"가","나"]
print(mylist)
```

```python
mylist.append("카")
print(mylist)
```

```python
mytuple = (1,2,3,"가","나")
print(mytuple)
```

```python
mydict = {"등수":1  ,  "성별":'2'  ,  '등급':'가등급'}
print(mydict)
```
