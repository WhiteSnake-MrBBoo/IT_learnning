## print ( 값 or 변수 , end="\n" )

````python
for v in range(1, 11):
    print(v * 10 , end="\t")
````

````python
for v in range(1,6) :
    print(v * 10, end="")
    if(v == 5) :
        print("", end="")
    else :
        print("-", end="")
````

````python
for number in range(1,11) :
    if(  number%2 == 0) :
        print(number, "짝", end="\t")
    else :     
        print(number, "홀", end="\t")
````

* 1부터 10까지 합

````python
total = 0

for number in range(1,11) :
    total = total + number

print(total)
````

* 1부터 10까지 짝수 합

````python
total = 0

for number in range(1,11) :
    if  (number%2 == 0) :
        total = total + number

print(total)
````

* 리스트에서 20 보다 작은 3의 배수를 출력하라
* mylist = [13, 21, 12, 14, 30, 18]

````python
mylist = [13, 21, 12, 14, 30, 18]
for x in mylist:
    if (x<20 and x%3==0) : 
        print(x)
````

* 구구단

````python
for dan in range(2,10) :
    #print(dan,"단")
    print(f"{dan}단")
````

````python
for gob in range(1,10) :
    print(f"*{gob}=", end="\t")
````

````python
for dan in range(2,10) :
    print(f"{dan}단", end="\t")
    for gob in range(1,10) :
        print(f"{dan}*{gob}={dan*gob}", end="\t")
    print()
````
