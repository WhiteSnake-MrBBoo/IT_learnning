# **requests 정적 크롤링**

```python
from bs4 import BeautifulSoup
import requests


url = "https://star.ytn.co.kr/news/list.php?mcd=0117&hcd=04"
res = requests.get(url)
print(res.status_code) 
#print(res.text) 

html_doc = res.text
soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.title)
```

```python
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")
for news in news_list:
    title = news.select_one("div.text_area > div.title > a").text
    rdate = news.select_one("div.text_area > div.info > div.date").text
    img   = news.select_one("div.photo > a > img").get("data-src")
    href   = news.select_one("div.text_area > div.title > a").get("href")
    print(title, rdate, img, href)
    
    
```

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from bs4 import BeautifulSoup
import time

driver = webdriver.Chrome()
url = "https://star.ytn.co.kr/news/list.php?mcd=0117&hcd=04"
driver.get(url)
time.sleep(2)

# 3. 더보기 버튼 3번 클릭
for i in range(3):
    btn = driver.find_element(By.CSS_SELECTOR,"a.btn_white_arr_down")
    btn.click()
    time.sleep(2)

# 4. 현재 화면의 HTML 가져오기
html = driver.page_source
soup = BeautifulSoup(html, "html.parser")
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")
news_list = soup.select("#container > div > div.content > div > div.news_list_wrap > div")

for news in news_list:
    title = news.select_one("div.text_area > div.title > a").text
    rdate = news.select_one("div.text_area > div.info > div.date").text
    img   = news.select_one("div.photo > a > img").get("data-src")
    href   = news.select_one("div.text_area > div.title > a").get("href")
    print(title, rdate, img, href)

driver.quit()
```

# BeautifulSoup

```python
from bs4 import BeautifulSoup
import requests


url = "https://www.bucheon.go.kr/site/program/board/photoboard/list?boardtypeid=26778&menuid=148006001012001"
res = requests.get(url)
print(res.status_code) 
#print(res.text) 

html_doc = res.text
soup = BeautifulSoup(html_doc, 'html.parser')
print(soup.title)
```

```python
restaurant_list = soup.select("#boardForm > div > ul > li > div")

for restaurant in restaurant_list:
    title = restaurant.select_one("div > a > img").get("alt")
    #title = food.select("a")[1].text
    img   = "https://www.bucheon.go.kr"+restaurant.select_one("div > a > img").get("src")
    area  = restaurant.select_one("p:nth-child(3)").text
    tel   = restaurant.select_one("p:nth-child(4)").text
    
    print(title, img, area, tel)
```

# selenium

```python
from bs4 import BeautifulSoup
from selenium import webdriver
from selenium.webdriver.common.by import By
import time

url = "https://www.bucheon.go.kr/site/program/board/photoboard/list?boardtypeid=26778&menuid=148006001012001"
driver = webdriver.Chrome()
driver.get(url)
time.sleep(2)


for page in range(1, 4):
    print(f"{page}---------------------------------------------")
    # 페이지 이동
    if page > 1:
        page_url = f"{url}&currentpage={page}"
        driver.get(page_url)
        time.sleep(2)
    html_doc = driver.page_source 

    #----------- 이하 코드 requests 복붙 -----------
    soup = BeautifulSoup(html_doc, "html.parser")
    food_list = soup.select("#boardForm > div > ul > li > div")
    for food in food_list:
        title = food.select_one("div > a > img").get("alt")
        #title = food.select("a")[1].text
        img = "https://www.bucheon.go.kr" + food.select_one("div > a > img").get("src")
        addr = food.select_one("p:nth-child(3)").text
        tel = food.select_one("p:nth-child(4)").text
        print(title, img, addr, tel)
        
    
driver.quit()
```
