# **제어문 : if**

* score = 90이상 이면 A, 80 이상 이면 B, 아니면 F 출력

````python
score = 95
if score >= 90 :
    print ("A")
elif score >= 80 :
    print ("B")
else :
    print ("F")
````

# **제어문 : while**

* **[암기]1부터 5까지 while문 사용해 출력**

````python
v = 1
while (True) :
    print(v)
    if (v == 5) : 
        break
    v = v + 1
````

````python
num = 1 #while은 항상 시작 변수가 있다.
while(True) : #일단 들어오고, 나갈 궁리를 해!
    if num > 5:
        break
        
    else : 
        print(num, end=" ")
        num = num + 1
````

````python
num = 1 #조건에 맞는 값만 들어와!
while(num <6) :
    print(num, end="\t")
    num = num + 1
    
````

* **숫자를 입력받아 10배 값으로 반환하는 함수**
    * 함수이름: calc_point
    * 파라미터: num
    * 기능: 입력값의 10배
    * 리턴: 반환값

````python
def calc_point(num) :
    res = num*10
    return res

a = calc_point(7)
print(a)
````

* **숫자 2개를 입력받아 합의 값으로 반환하는 함수**
    * 함수이름: calc_point2
    * 파라미터: num1, num2
    * 기능: num1 2배값과 num2 10배값의 합
    * 리턴: 계산값

````python
def calc_point2(num1, num2) :
    res = num1*2+num2*10
    return res

a = calc_point2(3,4)
print(a)
````
