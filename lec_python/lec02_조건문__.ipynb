## 멀티 조건문

* and : 둘 다 True

````python
age = 60
gen = 'm'
print( age < 30     and      gen =='m' )
````

* or : 둘 중 하나라도 True

````python
age = 60
gen = 'm'
print( age < 30      or       gen =='m' )
````

````python
# 30미만 남성 키는 150이상
age = 25
gen = 'f'
height = 178
print( age < 30   and   gen =='m'   and height>=150)
````

````python
age = 25
gen = 'm'
height = 178

if ( age < 30   and   gen =='m'   and height>=150) :
    print("군필")
else :
    print("미필")
````

* in

````python
print(  3 in [1,2,3] )
````

````python
mylist = [1,2,3]
3 in mylist
````

* not

````python
mylist = [1,2,3]
3 not in mylist
````

* ==, !=

````python
a = 10
b = 20
a == b 
````

````python
a = 10
b = 20
a != b 
````

````python
a = 10
b = 10

if (a != b ) :
    print("다르다")
else :
    print("같다")
````

## if문법

````python
score = 85
if (score >=90 ) :
    print("A")
````

````python
score = 85
if (score >=90 ) :
    print("A")
else :
    print("나머지")
````

* 90이상이면 A  , 80이상이면 B  ,  70이상이면 C  ,  나머지

````python
score = 85

if (score >=90 ) :
    print("A")    
elif (score >=80 ) :
    print("B")  
elif (score >=70 ) :
    print("C")    
else :
    print("나머지")

print("---done---")    
````
