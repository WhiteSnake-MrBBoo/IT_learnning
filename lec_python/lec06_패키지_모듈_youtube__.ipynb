# **[실습] 공공데이터포털**
* https://www.data.go.kr
* https://www.data.go.kr/data/15058601/openapi.do
* 김해시 공사 : https://www.gimhae.go.kr/00761/00832/05867.web
* pip install requests

````python
import requests

url = "http://www.gimhae.go.kr/openapi/tour/restaurant.do"

headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/151.0.0.0 Safari/537.36"
    ),
    "Accept": "application/json, text/javascript, */*; q=0.01",
    "Accept-Language": "ko-KR,ko;q=0.9,en-US;q=0.8,en;q=0.7",
    "Referer": "https://www.gimhae.go.kr/",
}


res = requests.get(
    url,
    headers=headers,
    timeout=10
)

print(res.status_code) #여기가 정상이라면 200이 온다고 함
data = res.json()
print(data['results'])
````

````python
print(data["results"][0]  ['name'])
print(data["results"][0]  ['category'])
print(data["results"][0]  ['phone'])
print(data["results"][0]  ['address'])
````

````python
for v in data["results"] :
    print(v['name'] , v['category'], v['phone'], v['address'], v['xposition'],  v['yposition']) 
    print("--"*20)
````
