```python
res = re.match('[a-z]' , "ABC")
res
```

```python
res = re.match('[a-z]' , "1abc")    #------첫글자가 매칭안되면 스탑~~~
res
```

```python
res = re.search('[a-z]' , "1abc")   #------첫글자가 매칭안되면 다음~문장끝 매칭되면 스탑
print( res,  res[0] )
```

```python
res = re.findall('[a-z]' , "1abc")  #------첫글자가 매칭안되면 다음~문장끝 매칭되도 끝까지 스탑 -- []
res
```

```python
resobj = re.finditer('[a-z]' , "1abc")  #------첫글자가 매칭안되면 다음~문장끝 매칭되도 끝까지 스탑 -- 객체
resobj
```

```python
for r in resobj:
    print(r,   r.group() ,   r[0])
```

## 이메일 찾기
* abc@aa.com 
* lee.kim@korea.com

[\w\.-]+@[\w\.-]+\.\w+

```python
p1 = re.compile("[\w\.-]+@[\w\.-]+\.\w+")
res = p1.match('abc@aa.com')
print(res)

p = re.compile('[a-z]*[.a-z]*@[A-Za-z]+\.[A-Za-z]{3,}')
res = p1.match('lee.kim@korea.com')
print(res)

p3 = re.compile('[\w.]+@[\w]+\.[\w]+')
res = p3.match('lee.kim@korea.com')
print(res)
```

## 전화번호
* 010-123-1234
* 010-4561-4545
* 02-745-7895
* 02-9258-7895

```python
p = re.compile('[\d]{2,3}-[\d]{3,4}-[\d]{4}')
res = p.match('010-4561-4545')
print(res)

p1 = re.compile(r"^0(2|10)\-[0-9]{3,4}\-[0-9]{4}$")
res = p1.match('010-4561-4545')
print(res)
```

## 호칭 찾기

```python
name1     = "raund, Mr. Owen Harris"
name2   = "Cumings, Mrs. John Bradley (Florence Briggs Thayer)"
name3 = "Heikkinen, Dr. Laina"
```

```python
p = re.compile("\,\s([A-Za-z]+\.)\s")

res = p.findall(name1)
print(res[0])

res = p.findall(name2)
print(res[0])

res = p.findall(name3)
print(res[0])
```
