```python
def myprint(arr):
    print(f"type:{type(arr)}")
    print("shape: {}, dimension: {}, dtype:{}".format(arr.shape, arr.ndim, arr.dtype))
    # print(f"dtype: {arr.dtype}")
    # print(f"shape: {arr.shape}")
    # print(f"dimension: {arr.ndim}")
    print(arr)
```

## Broadcasting over axis 0 with a 1-D array : 차원늘리기

세로 방향(over axis 0)으로 row를 복사해가면서 Braodcasting을 하는 예<br>
<img src="https://t1.daumcdn.net/cfile/tistory/2708D74858A062B836" width=600>

```python
# (2) Broadcasting using a 1-D array# Arithmetic operations between 2-D array and 1-D array # that is the same length as the row-length

b = np.arange(12).reshape((4, 3))
print(b.shape)
print(b)
print("----------")
c = np.array([0, 1, 2])
print(c.shape)
print(c)
print("----------")
# adding c (1-D array) row-wise to b (2-D array)
print(b + c)
```

## Broadcasting over axis 1 with a 2-D array : 차원늘리기

가로 방향(over axis 1)으로 column을 복사해가면서 broadcasting하는 예 <br>
<img src='https://t1.daumcdn.net/cfile/tistory/2265944958A062CF0D' width=600>

```python
# (3) Broadcasting over axis 1 of a 2-D array

b = np.arange(12).reshape((4, 3))
print(b.shape)
print(b)
print("----------")
c = np.array([0, 1, 2, 3]).reshape(4, 1)
print(c.shape)
print(c)
print("----------")
# adding c (1-D array) row-wise to b (2-D array)
print(b + c)
```

## (1) 두 배열을 왼쪽에서 오른쪽으로 붙이기    

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print( np.r_[a, b]                    )    # ( ) 를 사용하지 않고 [ ] 를 사용하는 것에 주의
print( np.hstack([a, b])              )
print( np.concatenate((a, b), axis=0) )
```

## (2) 두 배열을 위에서 아래로 붙이기   


```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print( np.r_[[a], [b]]                )    # a, b 배열을 [ ]을 사용해서 1-D 배열로 만든거 주의
print( np.vstack([a, b])              )

# 1D 배열은 "AxisError: axis 1 is out of bounds for array of dimension 1" --> 2D 이상 배열 사용할 것 
# print( np.concatenate((a, b), axis=1) )
```

```python
c = np.array([[0, 1, 2], [3, 4, 5,]])
d = np.array([[6, 7, 8], [9, 10, 11]])
print(c)
print(d)
print( np.concatenate((c, d), axis=1) )
```

## (3) 두 개의 1차원 배열을 칼럼으로 세로로 붙여서 2차원 배열 만들기<br>(Stack 1-D arrays as columns into a 2-D array)  

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])
print( np.c_[a, b]                          )    # ( ) 를 사용하지 않고 [ ] 를 사용하는 것에 주의
print( np.column_stack([a, b])              )

# 1D 배열은 "AxisError: axis 1 is out of bounds for array of dimension 1" --> 2D 이상 배열 사용할 것 
# print( np.concatenate((a.T, b.T), axis=1) )
```

```python
c = np.array([[0, 1, 2], [3, 4, 5,]])
d = np.array([[6, 7, 8], [9, 10, 11]])
print(c)
print(d)
print( np.concatenate((c.T, d.T), axis = 1) )
```
