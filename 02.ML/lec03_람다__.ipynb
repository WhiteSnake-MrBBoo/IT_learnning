# **컨프리헨션 & 람다**

* <b>for문 1~10 짝수만 출력

````python
for v in range(1,11):
    if (v%2 ==0) :
        print(v)
````

````python
[v for v in range(1,11) if (v%2 ==0)  ]
````

* <b>for문 1~5 3보다크면 값*2, 그렇지 않으면 값 출력

````python
for v in range(1,6):
    if (v>3) :
        print(v*2)
    else :
        print(v)
````

````python
[v*2 if (v>3)  else v for v in range(1,6)  ]
````

# **람다(lambda)**
* 익명(이름없는) 함수
* 1회용

* 함수이름 : add
* 파라미터 : 숫자2개 
* 합산 결과 리턴

* <b>람다 예제

````python
def ncheck(n) :
    print(n*2) 
````

````python
ncheck2 = lambda n : n*2
````

````python
ncheck(6)   
````

````python
ncheck2(6)
````

* <b>람다 예제

````python
def add(n1, n2) :
    res = n1 + n2
    return res

a= add(1, 2)    
print(a)
````

````python
add2 = lambda n1, n2 : n1 + n2

a= add2(1, 2)    
print(a)
````

* <b>람다 예제

````python
def check(n1) :
    res = 0
    if n1 > 10 : 
        res = n1 * 2
    else :
        res = n1 
    return res    

check(2)
````

````python
a = lambda n1 : n1 * 2 if n1 > 10  else n1 
a(2)
````

````python
a = lambda n1 : check(n1)
a(20)
````

# **DataFrame을 활용한 lambda**

````python
import numpy as np
import pandas as pd
````

````python
df = pd.DataFrame({
    "name": ["kim", "hong", "park"],
    "age": [20, 25, 30]
})
df
````

````python
df['age'] = df['age'] + 10
df
````

````python
df['age'] = df['age'].apply(lambda a : a+10)
df
````

````python
df['etc'] = df['age'].apply(lambda a : '성인' if a>=45  else '미성년')
df
````
