# **형상 관리(Configuration Management, SCM)**

> **형상 관리(Configuration Management, SCM)는 소프트웨어 개발 과정에서 소스 코드, 문서, 데이터 등 변경 사항을 체계적으로 추적하고 통제하는 관리 활동**<br>
> **Git** : 가장 널리 쓰이는 분산 버전 관리 시스템으로, GitHub 같은 플랫폼과 함께 사용<br>
> **SVN**(Subversion) : 중앙 집중형 방식으로 소스 변경 이력을 관리하던 전통적 도구

> **변경 추적** : 파일이나 코드가 바뀔 때 누가, 언제, 왜 수정했는지 기록<br>
**품질 유지** : 여러 사람이 동시에 개발할 때 발생할 수 있는 충돌이나 오류를 막아 제품의 일관성과 무결성 유지<br>
**버전 관리** : 문제가 생겼을 때 이전의 안정적인 상태로 복구

> **주요 관리 대상**<br>
    **소스 코드** : 프로그램의 실제 구동 <br>
    **텍스트개발 문서** :요구 사항 정의서, 설계서, API 명세서 등<br>
    **데이터 및 설정 파일** : 빌드 환경이나 배포에 필요한 설정 값

# **Git vs. GitHub**
* <font color=red><b>Git : 버전 관리 도구(Tool)</b></font>
* <font color=red><b>GitHub : Git 저장소를 온라인에서 관리하고 협업하는 서비스(Platform)</b></font>
    * <i>Platform이란 : 여러 기능과 서비스를 이용할 수 있는 기반이 되는 공간

| 구분         | Git   (Tool)                          | GitHub (Platform)                                      |
| ---------- | ------------------------------- | -------------------------------------------- |
| **정의**     | 분산 버전 관리 <font color=red><b>시스템                    | Git 저장소를 위한 협업 <font color=red><b>플랫폼                           |
| **주요 목적**  | <font color=red><b>코드의 버전 관리                       | <font color=red><b>코드 공유 및 협업                                   |
| **개발사/운영** | Git                             | GitHub                                       |
| **설치 필요**  | <font color=red><b>로컬 컴퓨터에 설치                      | <font color=red><b>웹 서비스 사용 가능                                  |
| **인터넷 필요** | 기본적인 Git 작업은 불필요                | 원격 저장소와 연동할 때 필요                             |
| **저장 위치**  | 주로 <font color=red><b>로컬 저장소                       | <font color=red><b>원격 저장소                                       |
| **대표 기능**  | commit, branch, merge, checkout | repository, pull request, issue, code review |
| **협업**     | 직접적인 협업 기능은 제한적                 | 여러 개발자가 함께 작업하기 편리                           |
| **관계**     | GitHub 없이도 사용 가능                | Git을 기반으로 사용                                 |
| **예시**     | `git commit`, `git branch`      | GitHub에 push, Pull Request 생성                |

<pre>

   ↓                                          ↓

   ↓                                          ↓

   ↓

   ↓

# **커맨드**

| 명령어           | 구분           | 간단 설명                   |
| ------------- | ------------ | ----------------------- |
| `git init`    | Git          | 현재 폴더를 Git 저장소로 생성      |
| `git clone`   | Git          | 원격 저장소의 프로젝트를 내 컴퓨터로 복사 |
| `git status`  | Git          | 현재 변경 상태 확인             |
| `git add`     | Git          | 변경된 파일을 스테이징 영역에 추가     |
| `git commit`  | Git          | 변경 내용을 버전으로 저장          |
| `git log`     | Git          | 커밋 이력 확인                |
| `git diff`    | Git          | 변경된 내용 비교               |
| `git branch`  | Git          | 브랜치 확인/생성               |
| `git switch`  | Git          | 다른 브랜치로 이동              |
| `git merge`   | Git          | 브랜치를 합치기                |
| `git remote`  | Git          | 원격 저장소 연결 정보 관리         |
| `git push`    | Git → GitHub | 로컬 변경 내용을 GitHub에 업로드   |
| `git pull`    | GitHub → Git | GitHub의 변경 내용을 가져와 반영   |
| `git fetch`   | GitHub → Git | GitHub의 변경 내용을 가져오기만 함  |
| `git restore` | Git          | 파일의 변경 내용을 되돌림          |
