```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()

#-------------------- 차트 관련 속성 (한글처리, 그리드) -----------
plt.rcParams['font.family']= 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

# 기본

```python
plt.figure(figsize=(10, 2)) 

plt.plot([1, 2, 3, 4], [1, 4,  9, 16], 'b--', label="블루")
plt.plot([1, 2, 3, 4], [2, 5, 10, 14], 'r--', label="레드")

plt.fill_between([2,3], [4,9], alpha=0.5)                           #----- 채우기

plt.axvline(1.5, 0.2, 0.8           , linestyle='-', linewidth=2)   #----- 수직선
plt.vlines(1.8, 5, 10, color='red'  , linestyle='-', linewidth=3)   #----- 수직선
plt.hlines(10, 1, 4,   color='green', linestyle='-', linewidth=3)   #----- 수평선

plt.title("차트제목")                 #-------- 제목
plt.xlabel('X-한글')                  #-------- 레이블
plt.ylabel('Y-Label')
plt.xlim(0,6)                         #-------- 눈금
plt.ylim(0,20)
plt.grid(True)                        #-------- 그리드격자
plt.legend()  #loc="upper right")     #-------- 범례
plt.show()
```

# **plt.subplot<font color=red>s</font>() vs. plt.subplot()**

```python
# 1줄 3칸으로 ax로 나눠서 그림
fig, (ax1, ax2, ax3) = plt.subplots(1,3,     figsize=(14, 3)) 

ax1.bar([1, 2, 3, 4], [2, 5, 10, 14]     )
ax2.plot([1, 2, 3, 4], [1, 4,  9, 16]      , color='red')
ax3.bar([1, 2, 3, 4], [2, 5, 10, 14]     )

ax1.set_xlabel('X111-한글')                  #-------- ax개별 레이블
ax1.set_title('X111-차트제목')                #-------- ax개별 타이블

ax2.set_xlabel('X222-한글')                  #-------- ax개별 레이블
ax2.set_title('X222-차트제목')                #-------- ax개별 타이블

ax3.set_xlabel('X333-한글')                  #-------- ax개별 레이블

fig.suptitle('전체 차트 제목')

plt.tight_layout()
plt.show()
```

```python

####fig, (ax1, ax2, ax3) = plt.subplots(1,3,     figsize=(14, 3)) 
plt.figure(figsize=(10, 2)) 

plt.subplot(1,3,      1)         #---------------------------1행3열의 1번째
plt.bar([1, 2, 3, 4], [2, 5, 10, 14]     )

plt.subplot(1,3,      2)         #---------------------------1행3열의 2번째
plt.plot([1, 2, 3, 4], [1, 4,  9, 16]      , color='red')

plt.subplot(1,3,      3)         #---------------------------1행3열의 3번째
df = sns.load_dataset("fmri")
sns.barplot(x="timepoint", y="signal", data=df, hue="region", errorbar=None)  #errorbar=('ci', 95)

plt.show()
```

# 스캐터 : plt.scatter()

```python
import numpy as np

plt.figure(figsize=(10, 2)) 

np.random.seed(0)
n = 50
x = np.random.rand(n)
y = np.random.rand(n)

plt.scatter(x, y)
plt.show()
```

# 파이 : plt.pie()

```python
plt.figure(figsize=(10, 2))
ratio = [34, 32, 16, 18]
labels = ['Apple', 'Banana', 'Melon', 'Grapes']

plt.pie(ratio, labels=labels, autopct='%.1f%%')
plt.show()
```

# 박스플롯 : plt.boxplot()

```python
plt.figure(figsize=(10, 2))
np.random.seed(0)
data_a = np.random.normal(0, 2.0, 20)
data_a = np.append(data_a, 12)
print(data_a)

plt.boxplot(data_a)
plt.show()


# 사분위값 계산
min_value  = np.min(data_a)
q1         = np.percentile(data_a, 25)
median     = np.percentile(data_a, 50)
q3         = np.percentile(data_a, 75)
max_value  = np.max(data_a)

print("최솟값 :", min_value)
print("25% (Q1) :", q1)
print("50% (중앙값) :", median)
print("75% (Q3) :", q3)
print("최댓값 :", max_value)
```
