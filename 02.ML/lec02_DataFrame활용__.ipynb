# **stat.**
* 통계(statistics)

```python
emp.describe()
```

* <font color=red><b>★★JOB별 사원수 

```python
emp.groupby("JOB")['EMPNO'].count()
emp.groupby("JOB")['JOB'].count()
emp['JOB'].value_counts()
```

* <font color=red><b>★★각 부서(DEPTNO)별 JOB별 급여(SAL) 총합

```python
emp['SAL']
```

```python
emp.groupby(['DEPTNO','JOB'])    ['SAL'].sum()
```

```python
emp.groupby(['DEPTNO','JOB'])['SAL'].sum().reset_index()
```

* <font color=red><b>★★부서(DEPTNO)별 급여(SAL) 최소 / 최대

```python
###emp.groupby( ["DEPTNO"]) ['SAL'].min()
emp.groupby( ["DEPTNO"]) ['SAL'].agg( ['min','max'] )
```

* <font color=red><b>★★부서(DEPTNO)별 급여(SAL) 최소 / 입사일(HIREDATE) 최대

```python
emp.groupby( ["DEPTNO"]).agg( {'SAL':'min','HIREDATE':'max'} )
```
