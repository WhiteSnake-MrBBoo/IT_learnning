# **★★★Grouping★★★**
* max() , min(), mean(), sum(), count(), std()

```python
df77 = pd.DataFrame(
    {
        "A": ["one", "one", "two", "three"] * 3,
        "B": ["A", "B", "C"] * 4,
        "C": ["foo", "foo", "foo", "bar", "bar", "bar"] * 2,
        "D": np.random.randn(12),
        "E": np.random.randn(12),
    }
)
df77.head(3)
```

```python
df77.groupby("A")["D"].min()
```

```python
df77.groupby("A")["D"].agg(['min','max'])
```

```python
#df77.groupby("A")["D"].agg(  ['min','max']  )
df77.groupby("A")      .agg(  {"D":"min",   "E":"max"}  )
```

# **concat vs. merge**

## **pd.concat([adf, bdf],  axis=)**
* <font color=red><B>0 : 위(2) 아래(3)  = 총5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열

```python
dfa = pd.DataFrame({"a" : [1,2,3] , "b":[10,20,30]})
dfb = pd.DataFrame({"a" : [4,5]   , "b":[40,50]})
```

```python
dfa
```

```python
dfb
```

```python
dfab = pd.concat( [dfa, dfb], axis=0)
dfab = dfab.reset_index(drop=True)  #-----------인덱스 정리
dfab
```

```python
dfab = pd.concat( [dfa, dfb], axis=1)
dfab.columns = ['a','b','c','d']  #-----------컬럼 정리
dfab
```

## **pd.merge(adf, bdf  , on="컬럼", how="inner/outer")**
* <font color=red><B>0 : 위(2) 아래(3)  = 총5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열

```python
dfab = pd.merge(dfa,dfb, left_index=True, right_index=True, how='inner')
dfab
```

```python
dfab = pd.merge(dfa, dfb, left_index=True, right_index=True, how='outer')
dfab
```
