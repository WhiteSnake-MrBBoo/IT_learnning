* 위도 (Latitude): 약 37.4848°경도 (Longitude): 약 126.7831°

```python
import folium

m = folium.Map(location=(37.4848, 126.7831), zoom_start=15) #location = 지도 중심 위치 , zoom_start = 처음 확대 정도
#zoom  5  → 국가 수준,zoom  8  → 지역 수준,zoom 12  → 도시 수준,zoom 15  → 동네/거리 수준,zoom 18  → 건물 수준

folium.Marker(
    location=[37.4848, 126.7831],
    tooltip="풍선도움말",
    popup="<img width=100 height= 100 src='https://www.wooriac.co.kr/images/common/logo.svg'>우리인재개발원",
    icon=folium.Icon(color="green"),
).add_to(m)

folium.Marker(
    location=[37.484056, 126.782528],
    tooltip="부천역",
    popup="부천역...",
    icon=folium.Icon(color="red"),
).add_to(m)


folium.CircleMarker([37.484056, 126.782528], radius=100,color='#3186cc', fill_color='#3186cc', popup='부천역'
).add_to(m)
#radius → 화면 픽셀 크기 / folium.Circle( radius=100 -100m)

m
```

# **마커 클러스터**

```python
import numpy as np
from folium import plugins

N = 100 #100개의 좌표
data = np.array(
[
np.random.uniform(low=36, high=38, size=N), # Random latitudes. 36 ≤ 랜덤값 < 38 / N=100개 생성
np.random.uniform(low=127, high=128, size=N), # Random longitudes. 127 ≤ 랜덤값 < 128 / N=100개 생성
] ).T #Transpose, 전치

print(data[:3])
popups = [str(i) for i in range(N)] # N=0~99까지 100개 문자열 만들기 Popups texts are simple numbers.


m = folium.Map([37.566345, 126.977893], zoom_start=8)
plugins.MarkerCluster(data, popups=popups).add_to(m)
m
```
