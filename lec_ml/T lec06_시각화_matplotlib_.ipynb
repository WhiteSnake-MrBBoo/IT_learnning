```python
import matplotlib.pyplot as plt #별명으로 쓰는건 보통 파일이름!!
import seaborn as sns

#-------------------- 차트 관련 속성 (한글처리, 그리드) -----------
plt.rcParams['font.family']= 'Malgun Gothic'
plt.rcParams['axes.unicode_minus'] = False
```

# 기본

```python
plt.figure(figsize=(10, 2)) 

plt.plot([1, 2, 3, 4], [1, 4,  9, 16], 'b--', label="블루")
plt.plot([1, 2, 3, 4], [2, 5, 10, 14], 'r--', label="레드")

plt.fill_between([2,3], [4,9], alpha=0.5)
plt.axvline(1.5, 0.2, 0.8, color='lightgray', linestyle='-', linewidth=2) #-------수직선(비율로)
plt.vlines(1.8, 5, 10, color='lightgray', linestyle='solid', linewidth=3) #------------수직선(값으로)
plt.hlines(10, 1, 4, color='lightgreen', linestyle='solid', linewidth=3) #--------------수평선(값으로)

plt.title("차트제목")                 #-------- 제목
plt.xlabel('X-한글')                  #-------- 레이블
plt.ylabel('Y-Label')
plt.xlim(0,6)                         #-------- 눈금
plt.ylim(0,20)
plt.grid(True)                        #-------- 그리드격자
plt.legend()  #loc="upper right")     #-------- 범례
plt.show()
```

# plt.subplots() vs. plt.subplot()

```python
# 1줄 3칸으로 ax로 나눠서 그림
fig, (ax1, ax2, ax3 ) = plt.subplots(1, 3, figsize=(14, 3))

ax1.bar([1, 2, 3, 4], [1, 4,  9, 16])
ax2.plot([1, 2, 3, 4], [1, 4,  9, 16])
ax3.bar([1, 2, 3, 4], [1, 4,  9, 16])

###plt.xlabel("x111-한글")
ax1.set_xlabel("x111-한글")
ax1.set_title("x111-차트제목")

ax2.set_xlabel("x222-한글")
ax2.set_title("x222-차트제목")

ax3.set_xlabel("x333-한글")
ax3.set_title("x222-차트제목")

fig.suptitle('이건 전체 차트 제목이야')

plt.tight_layout()
plt.show()
```

```python
###fig, (ax1, ax2, ax3 ) = plt.subplots(1, 3, figsize=(14, 3))
plt.figure(figsize=(10,2))

plt.subplot(1,3,     1) #-------------------(행, 열, 차트번째)
plt.plot([1, 2, 3, 4], [1, 4,  9, 16] , color="red")

plt.subplot(1,3,     2) #-------------------(행, 열, 차트번째)
plt.bar([1, 2, 3, 4], [1, 4,  9, 16] , color="red")

plt.subplot(1,3,     3) #-------------------(행, 열, 차트번째)

df = sns.load_dataset("fmri")
sns.barplot(x="timepoint", y="signal", data=df, hue="region", errorbar=None)

plt.show
```

# 파이 : plt.ple()

```python
plt.figure(figsize=(10,2))
ratio = [34, 32, 16, 18]
labels = ['Apple', 'Banana', 'Melon', 'Grapes']

plt.pie(ratio, labels=labels, autopct='%.1f%%')
plt.show()
```

# 박스플롯 : plt.boxplot()

```python
plt.figure(figsize=(10,2))
np.random.seed(0)
data_a = np.random.normal(0, 2.0, 50)
data_a = np.append(data_a,12)
print(data_a)

plt.boxplot(data_a)
plt.show()

# 사분위값 계산
min_value = np.min(data_a)
q1        = np.percentile(data_a, 25)
median    = np.percentile(data_a, 50)
q3        = np.percentile(data_a, 75)
max_value = np.max(data_a)

print("최솟값 :", min_value)
print("25% (Q1) :", q1)
print("50% (중앙값) :", median)
print("75% (Q3) :", q3)
print("최댓값 :", max_value)
```
