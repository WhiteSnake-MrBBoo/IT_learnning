## mypred() 함수 : 공모전 test 예측

````python
def myscore222(df, model_r=None, model_c=None, input_cols=[]):

    # 7 : 3 분리
    train = df[df['dd'] <= 15]
    val   = df[df['dd'] > 15]

    # 학습 데이터
    train_X   = train[input_cols]
    train_y_r = train['registered']
    train_y_c = train['casual']

    # 검증 데이터
    val_X   = val[input_cols]
    val_y_r = val['registered']
    val_y_c = val['casual']

    if model_r is None:
        model_r = RandomForestRegressor(
            n_estimators=100,
            random_state=1234
        )

    if model_c is None:
        model_c = RandomForestRegressor(
            n_estimators=100,
            random_state=1234
        )

    print(model_r.__class__.__name__)
    print(model_c.__class__.__name__)

    # registered 학습
    model_r.fit(train_X, train_y_r)

    pred_r_log = model_r.predict(val_X)
    pred_r = np.expm1(pred_r_log)

    # casual 학습
    model_c.fit(train_X, train_y_c)

    pred_c_log = model_c.predict(val_X)
    pred_c = np.expm1(pred_c_log)

    pred = np.round(pred_r + pred_c)
    pred = np.maximum(pred, 0)

    y_test3 = (
        np.expm1(val['registered'])
        +
        np.expm1(val['casual'])
    )

    rmsle = root_mean_squared_log_error(
        y_test3,
        pred
    )

    print(f"rmsle : {rmsle:.4f}")

    return model_r, model_c
````

# **[6]전처리 & 가공 : 파생피쳐**
* peak, ideal, sticky

````python
for i, df in enumerate([train, test]):
    df['peak'] = df[['hh', 'workingday']].apply(
        lambda x: 1 if (
            (x['workingday'] == 1 and ( x['hh'] == 8 or 17 <= x['hh'] <= 18 or 12 <= x['hh'] <= 13))
            or
            (x['workingday'] == 0 and (10 <= x['hh'] <= 19) )
        ) else 0,
        axis=1
    )
    
    df['ideal'] = df[['temp', 'windspeed']].apply(
        lambda x: 1 if x['temp'] > 27 and x['windspeed'] < 30 else 0,
        axis=1
    )
    
    df['sticky'] = df[['humidity', 'workingday']].apply(
        lambda x: 1 if x['workingday'] == 1 and x['humidity'] >= 60 else 0,
        axis=1
    )
    
    if i == 0:
        train = df
    elif i == 1:
        test = df
        
````

## **GradientBoosting**
* Gradient Boosting(그래디언트 부스팅)은 여러 개의 단순하고 약한 예측 모델(주로 의사결정 트리)을 순차적으로 결합해 강력한 예측 성능을 만드는 앙상블(Ensemble) 학습 알고리즘

````python
from sklearn.base import clone

params = {
    'n_estimators': 150,
    'max_depth': 5,
    'random_state': 0,
    'min_samples_leaf': 10,
    'learning_rate': 0.1,
    'subsample': 0.7,
    'loss': 'squared_error'
}

gbm_model = GradientBoostingRegressor(**params)

gbm_cols = [
    'weather', 'temp', 'humidity', 'windspeed',
    'holiday', 'workingday', 'season',
    'hh', 'week', 'yy', 'ideal'
]

# registered / casual 용 모델을 각각 독립적으로 복제
gb_r, gb_c = myscore222(
    train,
    model_r=clone(gbm_model),
    model_c=clone(gbm_model),
    input_cols=gbm_cols
)
````
