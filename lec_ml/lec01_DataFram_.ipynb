# **★★★Grouping★★★**
* max(), min(), mean(), sum(), count(), std()

```python
df77 = pd.DataFrame(
    {
        "A": ["one", "one", "two", "three"] * 3,
        "B": ["A", "B", "C"] * 4,
        "C": ["foo", "foo", "foo", "bar", "bar", "bar"] * 2,
        "D": np.random.randn(12),
        "E": np.random.randn(12),
    }
)
df77.head(3)
```

```python
df77.groupby("A")["D"].min()
```

```python
df77.groupby("A")["D"].agg(["min","max"]) #[],"" 까먹지 말기
```

```python
#df77.groupby("A")["D"].agg(["min","max"])
df77.groupby("A").agg({"D":"min","E":"max"}) #list가 dict로 바뀐다!
```

# **concat vs. merge**

## concat(axis=)

* <font color=red><B>0 : 위(2) 아래(3)  = 총 5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열

```python
dfa = pd.DataFrame({"a" : [1,2,3] , "b":[10,20,30]})
dfb = pd.DataFrame({"a" : [4,5] , "b":[40,50]})
dfa
```

```python
dfa
```

```python
dfb
```

```python
# dfa.concat(dfb, axis=0)
pd.concat([dfa, dfb]) # ->이거 쓰셈
```

```python
dfab = pd.concat([dfa, dfb], axis=0)
dfab = dfab.reset_index(drop=True) #인덱스 정리하는 법 -> 기존 인덱스를 바깥으로 꺼내고, 컬럼이 추가 됨
                                    #drop=True 하면 기존 인덱스 버려짐
dfab
```

```python
dfab = pd.concat([dfa, dfb], axis=1)
dfab.columns=["a","b","c","d"]
dfab
```

## **merge(on="컬럼", how="inner/outer")**

* <font color=red><B>0 : 위(2) 아래(3)  = 총 5줄
* <font color=red><B>1 : 열(2) 열(3)    = 총5열

```python
dfab = pd.merge(dfa,dfb, left_index=True, right_index=True, how="inner")
dfab
```

```python
dfab=dfa.merge(dfb, left_index=True, right_index=True, how="outer")
dfab
```

```python
left = pd.DataFrame({"key": ["foo", "foo2"], "lval": [1, 2]})
right = pd.DataFrame({"key": ["foo", "foo2"], "rval": [4, 5]})
```

```python
left
```

```python
right
```

```python
dfab=left.merge(right, on="key", how="inner")
dfab
```
