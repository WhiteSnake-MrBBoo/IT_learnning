# **컴프리헨션 & 람다**

* **for문 1~10 짝수만 출력**

```python
for v in range(1,11) :
    if v % 2 == 0 :
        print(v, end=" ")
```

```python
[v for v in range(1,11)if (v % 2 == 0)] #다 한 줄에 나열하고, 내가 찾을 값을 맨 앞으로 보낸다.
```

* <b>for문 1~5 3보다크면 *2, 그렇지 않으면 값 출력

```python
for v in range(1,6):
    if v >3:
        print(v*2)
    else :
        print(v)
```

```python
[(v*2) if (v>3) else (v) for v in range(1,6)]
```

# **람다(Lambda)**
* 익명(이름없는) 함수
* 1회용

* <b>함수이름: add
* 파라미터: 숫자2개 
* 결과 합리턴

```python
def add(a,b) :
    res = a+b
    return res

c = add(4, 5)
print(c)
```

```python
def ncheck(n) :
    print(n*2)

ncheck(6)
```

* <b> 람다 예제

```python
ncheck2 = lambda n : n*2
```

```python
ncheck(6)
```

```python
ncheck2(6)
```

* <b> 람다 예제

```python
def add(a,b) :
    res = a+b
    return res

c = add(4, 5)
print(c)
```

```python
add2 = lambda a,b : a+b
print(add2(1, 3)) #print 해도 되고 안해도 되고~ "a="으로 바꿔도 되고 안바꿔도 되고~
```

# **DataFrame을 활용한 Lambda**

```python
import numpy  as np 
import pandas as pd
```

```python
df = pd.DataFrame({
    "name": ["kim", "hong", "park"],
    "age": [20, 25, 30]
})
df
```

```python
df["age"] = df["age"]+10
df
```

```python
df["age"] = df["age"].apply(lambda a: a+10)
df
```
