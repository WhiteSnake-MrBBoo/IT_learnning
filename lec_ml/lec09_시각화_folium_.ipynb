* 위도 (Latitude): 약 37.4848°
* 경도 (Longitude): 약 126.7831°

```python
import folium
m = folium.Map(location=(37.4848, 126.7831), zoom_start=12)

folium.Marker(
    location=[37.4848, 126.7831],
    tooltip="Click me!",
    popup="<img width=100 height= 100 src='https://www.wooriac.co.kr/images/common/logo.svg'>우리인재개발원",
    icon=folium.Icon(color="pink"),).add_to(m)

folium.Marker(
    location=[37.484056, 126.782528],
    tooltip="부천역",
    popup="부천역...",
    icon=folium.Icon(color="pink"),).add_to(m)

folium.CircleMarker([33.510619228109334, 126.49137485480217], radius=100,color='#3186cc', fill_color='#3186cc', popup='부천역').add_to(m)

m
```
